# Topic 1: SmartLend Initial EDA

Use this notebook to load the Give Me Some Credit dataset, inspect data quality, and produce the plots required in the lab.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
```

```python
DATA_PATH = "../data/raw/cs-training.csv"
df = pd.read_csv(DATA_PATH, index_col=0)

print(f"Shape: {df.shape}")
display(df.head())
display(df.dtypes)
```

## Data Quality Notes

Record missing values, target imbalance, unusual ranges, and anything that may affect preprocessing decisions in Topic 2.

```python
display(df.describe())
print("Missing values per column:")
display(df.isnull().sum())
print("Target distribution:")
display(df["SeriousDlqin2yrs"].value_counts())
display(df["SeriousDlqin2yrs"].value_counts(normalize=True).round(3))
```

## Target Distribution

```python
fig, ax = plt.subplots(figsize=(6, 4))
df["SeriousDlqin2yrs"].value_counts().plot(kind="bar", ax=ax, color=["steelblue", "coral"])
ax.set_title("Target Class Distribution: SeriousDlqin2yrs")
ax.set_xlabel("Class (0 = No Default, 1 = Default)")
ax.set_ylabel("Count")
ax.tick_params(axis="x", rotation=0)
plt.tight_layout()
plt.savefig("../docs/eda_target_distribution.png", dpi=150)
plt.show()
```

## Task 8 Observations

The target variable `SeriousDlqin2yrs` is highly imbalanced. Approximately 93% of applicants belong to class 0, only around 7% belong to class 1.

This imbalance is important because a model could achieve high accuracy by predicting class 0 for every applicant, but such a model would fail to identify many high-risk customers. Therefore, evaluation metrics such as precision, recall and F1-score will be more useful than accuracy alone when assessing model performance.

## Age and Income Distributions

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df["age"].plot(kind="hist", bins=40, ax=axes[0], color="steelblue", edgecolor="white")
axes[0].set_title("Age Distribution")
axes[0].set_xlabel("Age")

df["MonthlyIncome"].dropna().clip(upper=20000).plot(kind="hist", bins=50, ax=axes[1], color="coral", edgecolor="white")
axes[1].set_title("Monthly Income Distribution (clipped at 20000)")
axes[1].set_xlabel("Monthly Income")

plt.tight_layout()
plt.savefig("../docs/eda_age_income.png", dpi=150)
plt.show()
```

## Task 9 Observations

Most applicants are middle-aged, with the age distribution centred around the late 40s to early 50s.

Monthly income is right-skewed, meaning most applicants have lower to moderate incomes while a small number earn significantly more.

This dataset is probably not representative of the entire population because it focuses on people involved in credit and lending rather than all individuals.

## Ratio Features

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
df["DebtRatio"].clip(upper=5).plot(kind="hist", bins=50, ax=axes[0], color="steelblue", edgecolor="white")
axes[0].set_title("Debt Ratio (clipped at 5)")
axes[0].set_xlabel("Debt Ratio")

df["RevolvingUtilizationOfUnsecuredLines"].clip(upper=2).plot(kind="hist", bins=50, ax=axes[1], color="coral", edgecolor="white")
axes[1].set_title("Revolving Utilisation (clipped at 2)")
axes[1].set_xlabel("Utilisation Rate")

plt.tight_layout()
plt.savefig("../docs/eda_ratios.png", dpi=150)
plt.show()
```

## Task 10 Observations

Both DebtRatio and RevolvingUtilizationOfUnsecuredLines contain extreme outliers. Several values exceed financially realistic ranges, suggesting possible data quality issues or unusual customer behaviour. These outliers may influence model training.

## Correlation Overview

```python
fig, ax = plt.subplots(figsize=(10, 8))
corr = df.corr()
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0, square=True, linewidths=0.5, ax=ax)
ax.set_title("Feature Correlation Matrix")
plt.tight_layout()
plt.savefig("../docs/eda_correlation.png", dpi=150)
plt.show()
```

## Task 11 Observations

The strongest positive correlations with `SeriousDlqin2yrs` are the past-due payment features (`NumberOfTimes90DaysLate`, `NumberOfTime30-59DaysPastDueNotWorse`, and `NumberOfTime60-89DaysPastDueNotWorse`). This suggests that previous missed payments are strong indicators of future delinquency.

There are also very strong correlations between the three past-due variables themselves (0.98-0.99), indicating potential multicollinearity. Most other features show relatively weak linear relationships with the target variable.
